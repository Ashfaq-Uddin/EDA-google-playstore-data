### RATING

```python
df['Rating'].unique()
```

```python
df['Rating'].describe()
```

```python
original_std = df['Rating'].std()
original_var = df['Rating'].var()
```

```python
corr_before = df.dropna(subset=['Rating']).corr()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(corr_before, annot=True, fmt=".2f", cmap="coolwarm", square=True)
plt.title("Correlation Matrix Before Imputation")
plt.show()
```

```python
rating_median = df['Rating'].median()
df['Rating'] = df['Rating'].fillna(rating_median)
```

```python
new_std = df['Rating'].std()
new_var = df['Rating'].var()
```

```python
print("Standard Deviation Before:", original_std)
print("Standard Deviation After:", new_std)
print("Variance Before:", original_var)
print("Variance After:", new_var)
```

```python
corr_after = df.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_after, annot=True, fmt=".2f", cmap="coolwarm", square=True)
plt.title("Correlation Matrix After Imputation")
plt.show()
```

```python
print("\nCorrelation difference (absolute):")
print((corr_before - corr_after).abs())
```

```python
numeric_features = [feature for feature in df.columns if df[feature].dtype != 'O']
categorical_features = [feature for feature in df.columns if df[feature].dtype == 'O']


print('We have {} numerical features : {}'.format(len(numeric_features), numeric_features))
print('We have {} categorical features : {}'.format(len(categorical_features), categorical_features))
```

### EDA

```python
plt.figure(figsize=(15,15))
plt.suptitle('Univariate Analysis of Numerical Features', fontsize=20, fontweight='bold', alpha=0.8, y=1.)

for i in range(0, len(numeric_features)):
    plt.subplot(4, 4, i + 1)
    sns.kdeplot(x=df[numeric_features[i]], shade=True, color='r')
    plt.xlabel(numeric_features[i])
    plt.tight_layout()
```

```python
plt.figure(figsize=(20, 15))
plt.suptitle('Univariate Analysis of Categorical Features', fontsize=20, fontweight='bold', alpha=0.8, y=1.)
category = ['Category','Android Ver','Genres']
for i in range(0, len(category)):
    plt.subplot(2,2,i+1)
    sns.countplot(x=df[category[i]], palette='Set2')
    plt.xlabel(category[i])
    plt.xticks(rotation=45)
    plt.tight_layout()
```

### Top ten APP Categories

```python
genre_category = pd.DataFrame(df['Genres'].value_counts())
genre_category.rename(columns = {'Genres':'Count'},inplace=True)
```

```python
plt.figure(figsize=(15,6))
sns.barplot(x=genre_category.index[:10], y = 'Count',data = genre_category[:10],palette='hls')
plt.title("Top 10 App genre")
plt.xticks(rotation=90)
plt.show()
```

```python
largest_installs_category = df.groupby('Category')['Installs'].sum().reset_index()
largest_installs_category = largest_installs_category.sort_values(by='Installs', ascending=False)
```

```python
plt.figure(figsize=(15, 6))
sns.barplot(x='Category', y='Installs', data=largest_installs_category, palette='viridis')
plt.title('Total Installs per Category', fontsize=16)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
top_categories = df.groupby('Category')['Installs'].sum().sort_values(ascending=False).head(5).index
```

```python
plt.figure(figsize=(20, 15))

# Step 2: Loop and plot top 5 apps in each top category
for i, cat in enumerate(top_categories):
    top_apps_df = df[df['Category'] == cat].sort_values(by='Installs', ascending=False).head(5)
    top_apps_df = top_apps_df[['App', 'Installs']]  # Just to make things clean

    plt.subplot(3, 2, i + 1)
    sns.barplot(x='Installs', y='App', data=top_apps_df, palette='coolwarm')
    plt.title(f'Top 5 Apps in {cat}', fontsize=14)
    plt.xlabel('Installs')
    plt.ylabel('App')

plt.tight_layout()
plt.show()
```
